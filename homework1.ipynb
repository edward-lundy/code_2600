```python
import pandas as pd
import numpy as np
Hitters = pd.read_csv('../data/Hitters.csv')


Hitters.shape
print(Hitters.median(numeric_only=True))

AtBat = Hitters["AtBat"].sum()
Outs = Hitters["AtBat"].sum() - Hitters['Hits'].sum()
Hits = Hitters['Hits'].sum()
Walks = Hitters['Walks'].sum()

print(AtBat)
print(Outs)
print(Hits)
print(Walks)

import matplotlib.pyplot as plt

pie_values = [Hits, Walks, Outs]
pie_labels = ["Hits", "Walks", "Outs"]

plt.pie(pie_values, labels=pie_labels)
plt.title("At Bats in 1986")
plt.show()

print(sum(Hitters['Salary'].isnull()))

print(Hitters[Hitters['Salary'].isnull()].describe().loc['mean'])
print(Hitters[Hitters['Salary'].notnull()].describe().loc['mean'])

plt.hist(Hitters[Hitters['Salary'].notnull()]['Salary'])
plt.title("Salary distribution on Opening Day 1987")
plt.ylabel("Number of Players")
plt.xlabel("Salary(Thousands)")
plt.show()

hitters_with_salary = Hitters[Hitters['Salary'].notnull()]
career_metrics = ['CAtBat', 'CHits', 'CHmRun', 'CRuns', 'CRBI', 'CWalks']
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 18))
for i, metric in enumerate(career_metrics):
    ax = axes.flatten()[i]
    ax.scatter(hitters_with_salary['Salary'], hitters_with_salary[metric])
    ax.set_title(f'Salary vs. {metric}') 
    ax.set_xlabel('Salary (in Thousands)')
    ax.set_ylabel(metric)
plt.tight_layout() 
plt.show()

hitters_with_salary.groupby('Division')['Salary'].mean()


players_with_12_years = hitters_with_salary[hitters_with_salary['Years'] > 12]
players_under_5_years = hitters_with_salary[hitters_with_salary['Years'] < 5]

data_to_plot = [players_with_12_years['Salary'], players_under_5_years['Salary']]

plt.boxplot(data_to_plot)
plt.title('Salary Distribution by Experience')
plt.ylabel('Salary (in Thousands)')
plt.xticks([1, 2], ['> 12 Years', '< 5 Years'])
plt.show()

Hitters['CAvg'] = Hitters['CHits'] / Hitters['CAtBat']
best_hitter_idx = Hitters['CAvg'].idxmax()
print(Hitters.loc[best_hitter_idx])

Hitters.sort_values(by="CAvg", ascending=False).head(5)
```
